**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
## 

samples = data.groupby("phenomenonTimeReferenceYear")["resultNumberOfSamples"].sum()
samples.plot(kind="bar", figsize=(9, 5), title="Reported Samples per Year")
plt.xlabel("Year")
plt.ylabel("Samples taken by Year")
plt.show()

copy = data.copy()

copy["resultNumberOfSamples"] = pd.to_numeric(
    copy["resultNumberOfSamples"], errors="coerce"
)
copy["property"] = copy["observedPropertyDeterminandLabel"].fillna(
    copy["observedPropertyDeterminandCode"]
)

top_properties = (
    copy.groupby("property")["resultNumberOfSamples"]
        .sum()
        .nlargest(8)
        .index
)

plot_data = copy[copy["property"].isin(top_properties)]

year_property = plot_data.pivot_table(
    index="phenomenonTimeReferenceYear",
    columns="property",
    values="resultNumberOfSamples",
    aggfunc="sum",
    fill_value=0
)

year_property.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6),
    colormap="tab20"
)

plt.xlabel("Year")
plt.ylabel("Reported number of samples")
plt.title("Reported Samples by Property and Year")
plt.legend(title="Property", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

Reporting was relatively steady from 1993 till 2003 and annualt totals  from2003 till 2012 were similar. From 2012 till 2014, the quantity of the measured samples increased 1,5 times. Looking into the properties of measured chemicals, then most measurements were made equally, as from the data that from 1992 till 2012 most measurements had proportionally the same quanitity. In 2013 and 2014, there were more Nitrate and Ammonium measurements taken compared to the other properties.
